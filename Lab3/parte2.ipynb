## Preguntas para reflexión:

### ¿Se puede identificar que los mensajes están cifrados con AES-CBC?

R: No directamente. Para un atacante que solo tenga acceso a los paquetes capturados por wireshark, el único indicio de que podriá tratarse de un cifrado mediante AES-CBC es la longitud del mensaje, que coincide con ser múltiplo de 16 como indica dicho algoritmo. Sin embargo, al no poder visualizar directamente la llave o el vector de inicialización, así como no poder identificar si el cifrado cambia para un mismo mensaje, no se puede concluir con certeza el algoritmo utilizado.

### ¿Cómo podríamos proteger más esta comunicación?

R: La comunicación actual ya permite ocultar de forma precisa los mensajes enviados. Sin embargo, si el atacante tiene certeza de que se está utilizando AES-CBC e identifica correctamente que los primeros 16 bytes del mensaje corresponden al vector de inicialización, puede iniciar a buscar patrones en los mensajes cifrados correspondientes a utilizar la misma llave para cada mensaje enviado. Para mejorar la seguridad de esta comunicación se podrían utilizar llaves aleatorias, o bien, una pareja de llave privada y pública a las que solo las personas involucradas en la comunicación tendrían acceso.
