### ¿Qué tamaño de clave se está usando para DES, 3DES y AES?

**R:** Se están utilizando los siguientes tamaños de clave para cada algoritmo implementado:
- DES: 8 bytes (64 bits. 56 efectivos y 8 de paridad).
- 3DES: 24 bytes (192 bits. 168 efectivos, 24 de paridad).
- AES: 16 bytes (128 bits). Se pudo haber utilizado también 24 o 32 bytes.

### ¿Qué modo de operación está implementado?

**R:** Se están implementando los modos ECB (Electronic Codebook) y CBC (Cipher Blocking Chaining). Con DES utilizamos ECB, con 3DES utilizamos CBC, y en el caso de AES utilizamos ambos modos de operación para realizar una comparación final con ambos resultados.

### ¿Por qué no debemos usar ECB en datos sensibles?

**R:** Como se puede apreciar en el resultado dentro de la imagen _Images/image_ecb.png_, utilizar ECB no nos garantiza ocultar por completo la información, puesto que no es capaz de ocultar patrones en los datos (en el caso de la imagen, representados como las siluetas de las figuras). Por lo tanto, utilizar ECB con datos sensibles puede no garantizar que un atacante no pueda acceder a dicha información.

### ¿Cual es la diferencia entre ECB vs CBC, se puede notar directamente en una imagen?

**R:** Sí. Como se mencionó en la pregunta anterior, ECB no es capaz de ocultar patrones en los datos, mientras que CBC sí. En las imágenes, podemos apreciar que en el cifrado con ECB aún es reconocible la silueta de la figura, mientras que en CBC se muestra como una mezcla homogénea de pixeles sin forma aparente. Esto evidencia la diferencia que existe entre ambos modos de operación, ya que, mientras ECB cifra cada bloque de manera independiente, utilizando solamente la llave indicada; para CBC cada bloque cifrado depende estrictamente de los bloques cifrados anteriores, iniciando con un vector de inicialización para el primer bloque a cifrar, lo cual garantiza que se oculta cualquier tipo de patrón en los datos originales.

### ¿Que es el IV?

**R:** El vector de inicialización (IV: Initialization Vector) es un bloque de bits aleatorios que se utiliza en conjunto con la llave indicada para cifrar el primer elemento de un conjunto de datos en modos de operación en cadena. Esto garantiza que mensajes idénticos cifrados con exactamente la misma clave, no produzcan el mismo texto cifrado, lo que resulta en un proceso de descifrado por fuerza bruta más complejo.

### ¿Que es el PADDING?

**R:** El padding o relleno es una técnica que garantiza que los datos a cifrar tengan una longitud que sea múltiplo exacto del tamaño de bloques a utilizar por el algoritmo. Es decir, garantiza que todos los bloques generados estén completos. Usualmente, al trabajar con datos binarios, se rellena con ceros (0) para no afectar el contenido original de los datos.

### ¿En qué situaciones se recomienda cada modo de operación?

**R:** ECB se recomienda utilizar únicamente en mensajes demasiado cortos y sin información sensible, o bien, para datos aleatorios que no puedan ser interpretados fácilmente aún si se llegaran a descifrar.
CBC, por otro lado, es recomendado para situaciones en las que se requiere un mayor nivel de confidencialidad de los datos. Es utilizado usualmente en el cifrado de archivos, bases de datos y comunicaciones seguras.

### ¿Cómo elegir un modo seguro en cada lenguaje de programación?

**R:** Para elegir un modo seguro de cifrado en cualquier lenguaje de programación, se deben utilizar bibliotecas criptográficas bien mantenidas y auditadas como PyCryptodome en Python, javax.crypto en Java, .NET Cryptography en C#, Web Crypto API en JavaScript, OpenSSL en C/C++, o el paquete crypto en Go. Se debe evitar completamente el modo ECB y utilizar CBC para uso general. Siempre se deben generar vectores de inicialización aleatorios para cada operación de cifrado y manejar correctamente el padding.
