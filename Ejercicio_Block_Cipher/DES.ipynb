```python
from Cryptodome.Cipher import DES
from Cryptodome.Random import get_random_bytes
import base64
```

## Generación una función cifrado y descifrado DES  
- Escriban una función en el lenguaje de programación que prefieran, que tome un mensaje en texto plano y lo cifre implementando el algoritmo de cifrado DES con el modo ECB.
- Implemente la generación aleatoria de la llave
- Implemente la función de relleno de bits manualmente

```python
# Funciones útiles
def getRandomKey(keySize=8):
    return get_random_bytes(keySize)
  
def manualPadding(text, blockSize=8):
    textBytes = text.encode('utf-8')
    paddingBytes = blockSize - (len(textBytes) % blockSize)
    
    padding = bytes([paddingBytes] * paddingBytes)
    return textBytes + padding
  
def removePadding(text, blockSize=8):
    paddingBytes = text[-1]
    
    if paddingBytes > blockSize:
        raise ValueError("Padding no válido")
    
    for i in range(1, paddingBytes + 1):
        if text[-i] != paddingBytes:
            raise ValueError("Padding no válido")
    
    return text[:-paddingBytes]
```

```python
# Cifrado
def cipherDES_ECB(text, key=None):
    if key is None:
        key = getRandomKey()
    
    fixedText = manualPadding(text)
    cipher = DES.new(key, DES.MODE_ECB)
    cipher_text = cipher.encrypt(fixedText)
    base64_text = base64.b64encode(cipher_text).decode('utf-8')
    
    return base64_text, key
```

```python
# Descifrado
def decipherDES_ECB(cipher_text, key=None):
    cipher_text = base64.b64decode(cipher_text)
    cipher = DES.new(key, DES.MODE_ECB)
    fixedText = cipher.decrypt(cipher_text)
    plainTextByte = removePadding(fixedText)
    plainText = plainTextByte.decode('utf-8')
    
    return plainText
```

```python
# Prueba Cifrado
message = ""

with open("Tests/des.txt", "r", encoding="utf-8") as file:
    message = file.read()

cipher, key = cipherDES_ECB(message)
print(f"Mensaje original:\n{message}\n")
print(f"Mensaje cifrado (base64):\n{cipher}\n")
print(f"Llave (hex): {key.hex()}")
```

```python
# Prueba Descifrado
decipher_message = decipherDES_ECB(cipher, key)
print(f"Mensaje descifrado:\n{decipher_message}")
```

```python
# Prueba unitaria
assert message == decipher_message, "¡Error! El mensaje descifrado no coincide con el original"
print("Prueba de cifrado y descifrado correcta")
```
