### Importación de librerías necesarias

```python
from Cryptodome.Cipher import DES3
from Cryptodome.Random import get_random_bytes
from Cryptodome.Util.Padding import pad, unpad
import base64
```

## Generación una función cifrado y descifrado 3DES
- Escriban una función en el lenguaje de programación que prefieran, que tome un mensaje en texto plano y lo cifre implementando el algoritmo de cifrado 3DES con el modo CBC.
- Implemente la generación aleatoria del la llave
- Utilice la función de relleno de bits de la librería por medio de pad

```python
# Funciones útiles
def getRandomKey(keySize=24):
    return get_random_bytes(keySize)
```

```python
# Cifrado
def cipher3DES_CCB(text, key=None):
    if key is None:
        key = getRandomKey()
    
    initVector = get_random_bytes(DES3.block_size)
    textBytes = text.encode('utf-8')
    
    paddingText = pad(textBytes, DES3.block_size)
    
    cipher = DES3.new(key, DES3.MODE_CBC, iv=initVector)  
  
    cipher_text = initVector + cipher.encrypt(paddingText)
    
    base64_text = base64.b64encode(cipher_text).decode('utf-8')
    
    return base64_text, key
```

```python
# Descifrado
def decipher3DES_CCB(cipher_text, key=None):
    data = base64.b64decode(cipher_text)
    initVector = data[:DES3.block_size]
    cipher_text = data[DES3.block_size:]
    cipher = DES3.new(key, DES3.MODE_CBC, iv=initVector)
    paddingText = cipher.decrypt(cipher_text)
    plainTextBytes = unpad(paddingText, DES3.block_size)
    plainText = plainTextBytes.decode('utf-8')
    return plainText
```

```python
# Prueba Cifrado
message = ""

with open("Tests/3des.txt", "r", encoding="utf-8") as file:
    message = file.read()

cipher, key = cipher3DES_CCB(message)
print(f"Mensaje original:\n{message}\n")
print(f"Mensaje cifrado (base64):\n{cipher}\n")
print(f"Llave (hex): {key.hex()}")
```

```python
# Prueba Descifrado
decipher_message = decipher3DES_CCB(cipher, key)
print(f"Mensaje descifrado:\n{decipher_message}")
```

```python
# Prueba unitaria
assert message == decipher_message, "¡Error! El mensaje descifrado no coincide con el original"
print("Prueba de cifrado y descifrado correcta")
```
